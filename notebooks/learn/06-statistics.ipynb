## 3. Effect size

Əvvəl kiçik və görünən nümunədə mexanizmi yoxlayın; sonra böyük layihə çıxışına keçin.

```python
effect=(a.mean()-b.mean())/np.sqrt((a.var(ddof=1)+b.var(ddof=1))/2);effect
```

## 4. Min p problemi

Əvvəl kiçik və görünən nümunədə mexanizmi yoxlayın; sonra böyük layihə çıxışına keçin.

```python
p=rng.uniform(size=1000);p.min(),(p<.05).sum()
```

**Özünü yoxla:** 1000 null testdə təxminən neçə nominal discovery gözlənir?

Əvvəl kiçik və görünən nümunədə mexanizmi yoxlayın; sonra böyük layihə çıxışına keçin.
